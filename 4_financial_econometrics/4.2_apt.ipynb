# Arbitrage Pricing Theory

Stephen Ross (1976) introduced APT because relying on a single market factor is physically unrealistic. APT states that an asset's returns are generated by a linear combination of multiple systemic risk factors, plus some idiosyncratic noise.

**The factor model setup.** Assume returns are generated by $k$ systematic factors:

$$
r_i = \alpha_i + \beta_{i1}f_1 + \beta_{i2}f_2 + \cdots + \beta_{ik}f_k + \epsilon_i
$$

where $f_1, \ldots, f_k$ are the common factors, $\beta_{ij}$ is asset $i$'s exposure (loading) to factor $j$, and $\epsilon_i$ is idiosyncratic noise — uncorrelated across assets and with the factors.

**The APT pricing relationship:**

$$
\mu_i - r_f = \beta_{i1}\lambda_1 + \beta_{i2}\lambda_2 + \cdots + \beta_{ik}\lambda_k
$$

Expected excess return is a linear combination of factor exposures, each weighted by that factor's **risk premium** $\lambda_j$ (the compensation for bearing one unit of exposure to factor $j$). CAPM is the $k=1$ special case: one factor (the market), one premium $\lambda_1 = \mu_m - r_f$, exposure $\beta_{i1} = \beta_i$.

The genius of APT isn't just adding more variables; it's the underlying proof. It doesn't require strict, unrealistic assumptions about utility curves or perfectly rational investors like CAPM does. Instead, it relies on one simple rule: The Law of One Price = no-arbitrage, not equilibrium. If an asset's actual return deviates from the return predicted by its factor exposures, an arbitrageur can construct a zero-cost, risk-free portfolio (by shorting the overvalued side and going long the undervalued side) to lock in a guaranteed profit. This arbitrage pressure forces the price back into alignment with the equation.

**Why this is more robust than CAPM:**

- No need to identify or observe the "true" market portfolio (CAPM's notorious weakness — Roll's critique: the market portfolio is unobservable, so CAPM is arguably untestable).
- No assumption that everyone has identical beliefs or mean-variance preferences.
- Allows multiple sources of priced risk, which matches what you see empirically — a single market factor explains only part of the cross-section of returns.
